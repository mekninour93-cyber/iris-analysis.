```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("iris.txt", sep=",")
```

```python
df.columns = ["sepal_length", "sepal_width", "petal_length", "petal_width", "species"]
```

```python
print("First 5 rows:\n", df.head())
```

```python
numeric_cols = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
```

```python
print("\nMaximum values:\n", df[numeric_cols].max())
print("\nMinimum values:\n", df[numeric_cols].min())
print("\nMean values:\n", df[numeric_cols].mean())
print("\nMedian values:\n", df[numeric_cols].median())
```

```python
grouped_stats = df.groupby("species")[numeric_cols].agg(["min", "max", "mean", "median"])
grouped_stats.columns = ['_'.join(col) for col in grouped_stats.columns]
print("\nStatistics by species:\n", grouped_stats)
```

```python
sns.countplot(x="species", data=df)
plt.title("Frequency Distribution of Iris Species")
plt.show()
```
